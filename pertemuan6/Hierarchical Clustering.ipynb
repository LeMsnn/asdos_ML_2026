```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage

X=[4.5,4,4.25,2.75,2,2.5]
Y=[3.4,4,4.25,3.25,3,2.5]

data = list(zip(X, Y))
print(data)
linkage_data = linkage(data, method='single', metric='euclidean')
dendrogram(linkage_data)
plt.title("Single Linkage")

#batas threshold misalkan = 1
max_d=1
plt.axhline(y=max_d, c='k')
plt.show()

linkage_data1 = linkage(data, method='complete', metric='euclidean')
plt.title("Complete Linkage")
dendrogram(linkage_data1)
plt.show()

linkage_data2 = linkage(data, method='average', metric='euclidean')
plt.title("Average Linkage")
dendrogram(linkage_data2)
plt.show()

linkage_data3 = linkage(data, method='ward', metric='euclidean')
plt.title("Ward Linkage")
dendrogram(linkage_data3)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering

asli={"x":[4.5,4,4.25,2.75,2,2.5],
    "y":[3.4,4,4.25,3.25,3,2.5]}

data=pd.DataFrame(asli)
print(data)
#data = list(zip(x, y))

#print(data)

linkage_data = linkage(data, method='single', metric='euclidean')
dendrogram(linkage_data)
#Threshold 
plt.axhline(y=max_d, c='k')
plt.show()

#default linkage ='ward'
hierarchical_cluster = AgglomerativeClustering(n_clusters=3, affinity='euclidean', linkage='single')
labels = hierarchical_cluster.fit_predict(data)
print(labels)

plt.scatter(data['x'], data['y'], c=labels)
plt.title("Agglomerative")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```
